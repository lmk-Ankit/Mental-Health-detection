```python
print('📂 Loading CSV (ZIP-aware)...')
print('=' * 70)


csv_path = _paths.input_path(
    os.path.join(MODEL_DIR, CSV_FILE) if not os.path.isabs(CSV_FILE) else CSV_FILE
)


def _resolve_csv(path):
    """If path is a ZIP, extract the first .csv inside and return its path."""
    import zipfile

    if zipfile.is_zipfile(path):
        print(f'   ℹ️  Detected ZIP archive: {os.path.basename(path)}')
        extract_dir = _paths.output_path(
            os.path.join(os.path.dirname(path), '_extracted_csv')
        )
        os.makedirs(extract_dir, exist_ok=True)
        with zipfile.ZipFile(_paths.input_path(path), 'r') as z:
            csv_names = [n for n in z.namelist() if n.lower().endswith('.csv')]
            if not csv_names:
                raise ValueError('❌ No .csv file found inside the ZIP.')
            z.extract(csv_names[0], extract_dir)
            extracted = os.path.join(extract_dir, csv_names[0])
            print(f'   ✅ Extracted: {csv_names[0]}')
            return extracted
    return path


csv_path = _paths.input_path(_resolve_csv(csv_path))


raw_df = None
for enc in ['utf-8', 'utf-8-sig', 'latin-1', 'cp1252']:
    try:
        raw_df = pd.read_csv(
            _paths.input_path(csv_path),
            encoding=enc,
            on_bad_lines='skip',
            engine='python',
        )
        print(f'   ✅ Encoding "{enc}" worked')
        break
    except UnicodeDecodeError:
        print(f'   ⚠️  "{enc}" failed, trying next...')

if raw_df is None:
    raise ValueError('❌ Could not read CSV.')

print(f'\n✅ Loaded  →  {raw_df.shape[0]} rows × {raw_df.shape[1]} columns')
for i, col in enumerate(raw_df.columns):
    print(f'   [{i:2d}] {col[:90]}')
raw_df.head(3)
```

```python
print('🔍 Extracting text — dual column layout...')
print('=' * 70)


def find_col_keyword(df, keyword, fallback_index=None):
    if fallback_index is not None:
        return df.columns[fallback_index]
    matches = [c for c in df.columns if keyword.lower() in c.lower()]
    return matches[0] if matches else None


col_a_q1 = find_col_keyword(raw_df, 'heaviest', Q1_COL_LAYOUT_A)
col_a_q2 = find_col_keyword(raw_df, 'alone', Q2_COL_LAYOUT_A)


col_b_rating = raw_df.columns[34] if raw_df.shape[1] > 34 else None
col_b_q1 = raw_df.columns[Q1_COL_LAYOUT_B or 35] if raw_df.shape[1] > 35 else None
col_b_q2 = raw_df.columns[Q2_COL_LAYOUT_B or 36] if raw_df.shape[1] > 36 else None


likert_cols = [raw_df.columns[j] for j in range(4, 34) if raw_df.shape[1] > j]

print(
    f'   Layout A — Q1: {str(col_a_q1)[:60]}'
    if col_a_q1
    else '   Layout A Q1 not found'
)
print(
    f'   Layout A — Q2: {str(col_a_q2)[:60]}'
    if col_a_q2
    else '   Layout A Q2 not found'
)
print(f'   Layout B — self-rating col: {col_b_rating}')
print(f'   Layout B — Q1 col: {col_b_q1}')
print(f'   Layout B — Q2 col: {col_b_q2}')
print(f'   Likert cols: {len(likert_cols)} found')


def likert_to_text(row, likert_cols):
    """Convert Likert frequency ratings into a descriptive text sentence.
    E.g. if 10 of 30 ratings are 'Very Often' or 'Often', we infer distress."""
    vals = [str(row[c]).strip() for c in likert_cols if pd.notna(row[c])]
    if not vals:
        return ''
    freq_map = {'Never': 0, 'Rarely': 1, 'Sometimes': 2, 'Often': 3, 'Very Often': 4}
    scores = [freq_map.get(v, 2) for v in vals]
    avg = sum(scores) / len(scores)
    often_count = sum(1 for v in vals if v in ('Often', 'Very Often'))
    if avg >= 3.0 or often_count >= 10:
        return (
            'I often feel overwhelmed, anxious or low. '
            'Many things bother me frequently. I struggle with my emotions often.'
        )
    elif avg >= 2.0 or often_count >= 5:
        return (
            'Sometimes I feel stressed or anxious. '
            'Some things bother me from time to time.'
        )
    else:
        return 'I mostly feel okay. Things are generally manageable for me.'


def extract_row_text(row):
    """Extract the best text from a row regardless of which layout was used."""
    parts = []

    if col_a_q1 and pd.notna(row[col_a_q1]):
        parts.append(str(row[col_a_q1]).strip())
    if col_a_q2 and pd.notna(row[col_a_q2]):
        parts.append(str(row[col_a_q2]).strip())

    if col_b_rating and pd.notna(row[col_b_rating]):

        sr = str(row[col_b_rating]).strip()
        if sr not in ('nan', ''):
            parts.append(f'Self-reported mood: {sr}.')
    if col_b_q1 and pd.notna(row[col_b_q1]):
        parts.append(str(row[col_b_q1]).strip())
    if col_b_q2 and pd.notna(row[col_b_q2]):
        parts.append(str(row[col_b_q2]).strip())

    if not parts and likert_cols:
        likert_text = likert_to_text(row, likert_cols)
        if likert_text:
            parts.append(likert_text)

    combined = ' '.join(parts).strip()
    return re.sub(r'\s+', ' ', combined)


df = pd.DataFrame()
df['text'] = raw_df.apply(extract_row_text, axis=1)
df['participant_id'] = [f'P{i+1:03d}' for i in range(len(df))]

ts_col = [c for c in raw_df.columns if 'timestamp' in c.lower()]
if ts_col:
    df['timestamp'] = pd.to_datetime(raw_df[ts_col[0]], errors='coerce')
    print(f'   ✅ Timestamp detected')

df = df[df['text'].str.len() > 10].reset_index(drop=True)
df['word_count'] = df['text'].apply(lambda x: len(x.split()))
n = len(df)

print(f'\n✅ {n} valid participants extracted')
print(
    f'   Word count: min={df["word_count"].min()}, max={df["word_count"].max()}, '
    f'mean={df["word_count"].mean():.0f}'
)
print('\n📋 Sample responses:')
for _, row in df.head(4).iterrows():
    print(f'  {row["participant_id"]}: {row["text"][:100]}...')
```

```python
from transformers import pipeline
import torch, re

print('🤖 Loading zero-shot classifier...')
zs_classifier = pipeline(
    'zero-shot-classification',
    model='facebook/bart-large-mnli',
    device=0 if torch.cuda.is_available() else -1,
)
print('   ✅ facebook/bart-large-mnli loaded')


VALID_CONDITIONS = [
    'addiction',
    'adhd',
    'alcoholism',
    'anxiety',
    'autism',
    'bipolarreddit',
    'bpd',
    'depression',
    'healthanxiety',
    'lonely',
    'ptsd',
    'schizophrenia',
    'socialanxiety',
    'suicidewatch',
]


ZS_LABELS = [
    'feeling persistently sad, empty, or hopeless with no energy',
    'excessive worry, overthinking, or constant anxiety and stress',
    'suicidal thoughts, wanting to die, or feeling life is not worth living',
    'attention difficulties, impulsivity, restlessness, or ADHD symptoms',
    'extreme mood swings between very high energy and very low crashes',
    'emotional instability, fear of abandonment, or identity confusion',
    'trauma, flashbacks, intrusive memories, or past abuse',
    'fear of social situations, embarrassment, or being judged by others',
    'excessive worry about physical health or body symptoms',
    'deep loneliness, social isolation, or feeling disconnected',
    'paranoia, hallucinations, or losing touch with reality',
    'alcohol misuse or dependency',
    'drug addiction or substance misuse',
    'difficulty understanding social cues or autism spectrum traits',
]
ZS_TO_CONDITION = {
    'feeling persistently sad, empty, or hopeless with no energy': 'depression',
    'excessive worry, overthinking, or constant anxiety and stress': 'anxiety',
    'suicidal thoughts, wanting to die, or feeling life is not worth living': 'suicidewatch',
    'attention difficulties, impulsivity, restlessness, or ADHD symptoms': 'adhd',
    'extreme mood swings between very high energy and very low crashes': 'bipolarreddit',
    'emotional instability, fear of abandonment, or identity confusion': 'bpd',
    'trauma, flashbacks, intrusive memories, or past abuse': 'ptsd',
    'fear of social situations, embarrassment, or being judged by others': 'socialanxiety',
    'excessive worry about physical health or body symptoms': 'healthanxiety',
    'deep loneliness, social isolation, or feeling disconnected': 'lonely',
    'paranoia, hallucinations, or losing touch with reality': 'schizophrenia',
    'alcohol misuse or dependency': 'alcoholism',
    'drug addiction or substance misuse': 'addiction',
    'difficulty understanding social cues or autism spectrum traits': 'autism',
}


REDDIT_FREQ = {
    'depression': 0.30,
    'suicidewatch': 0.12,
    'mentalhealth': 0.10,
    'anxiety': 0.07,
    'adhd': 0.06,
    'lonely': 0.06,
    'socialanxiety': 0.05,
    'bipolarreddit': 0.05,
    'bpd': 0.04,
    'ptsd': 0.04,
    'healthanxiety': 0.03,
    'autism': 0.03,
    'schizophrenia': 0.02,
    'alcoholism': 0.02,
    'addiction': 0.01,
}
UNIFORM = 1.0 / len(le.classes_)
CORRECTION = np.array([UNIFORM / REDDIT_FREQ.get(c, UNIFORM) for c in le.classes_])


KEYWORD_RULES = [
    (
        {
            'should i live',
            'want to die',
            'beauty of death',
            'loving death',
            'live or die',
            'end my life',
            'no reason to live',
            'suicide',
            'suicidal',
            'kill myself',
            'better off dead',
        },
        'suicidewatch',
        5.0,
    ),
    (
        {
            'flashback',
            'trauma',
            'intrusive',
            'haunting',
            'cant forget',
            'keeps coming back',
            'replaying',
            'past abuse',
            'triggered',
        },
        'ptsd',
        3.0,
    ),
    (
        {
            'mood swings',
            'highs and lows',
            'one minute fine',
            'next minute',
            'energy then crash',
            'rollercoaster',
            'which version of me',
            'wake up not knowing',
            'sudden crash',
            'crash down',
        },
        'bipolarreddit',
        3.0,
    ),
    (
        {
            'two people',
            'two versions',
            'split',
            'identity',
            'dont know who i am',
            'fear of abandonment',
            'push and pull',
        },
        'bpd',
        3.0,
    ),
    (
        {
            'worthless',
            'hopeless',
            'empty inside',
            'no energy',
            'cant get out of bed',
            'lost myself',
            'good for nothing',
            'useless',
            'numb',
            'void',
            'pointless',
            'heavy inside',
            'feel nothing',
        },
        'depression',
        2.5,
    ),
    (
        {
            'overthinking',
            'cant stop thinking',
            'racing mind',
            'what if',
            'constant worry',
            'always anxious',
            'panic',
            'heart racing',
        },
        'anxiety',
        2.0,
    ),
    (
        {
            'no one understands',
            'feel alone',
            'nobody cares',
            'completely alone',
            'disconnected',
            'no friends',
            'lost friends',
        },
        'lonely',
        2.0,
    ),
]


NORMAL_INDICATORS = {
    'peaceful',
    'content',
    'grateful',
    'enjoying',
    'happy',
    'positive',
    'calm',
    'balanced',
    'thriving',
    'fulfilled',
}


DISTRESS_INDICATORS = {
    'worthless',
    'hopeless',
    'empty',
    'numb',
    'hurt',
    'useless',
    'depressed',
    'anxious',
    'panic',
    'trauma',
    'abuse',
    'exhausted',
    'stuck',
    'trapped',
    'cry',
    'dark',
    'overwhelmed',
    'breakdown',
    'scared',
    'helpless',
    'heavy',
    'burden',
    'die',
    'death',
    'suicidal',
    'flashback',
    'intrusive',
    'crash',
    'void',
    'lost myself',
    'lonely',
}


def run_your_model(texts):
    """Reddit-trained ensemble with bias correction. Returns (N, 15) array."""
    X = tfidf.transform(texts)
    probas = []
    for model in models.values():
        if hasattr(model, 'predict_proba'):
            p = model.predict_proba(X)
        else:
            d = model.decision_function(X)
            p = np.exp(d - d.max(axis=1, keepdims=True))
            p /= p.sum(axis=1, keepdims=True)
        probas.append(p)
    avg = np.stack(probas).mean(axis=0) * CORRECTION
    return avg / avg.sum(axis=1, keepdims=True)


def run_zeroshot(text):
    """BART zero-shot. Returns dict of condition→probability. No mentalhealth."""
    result = zs_classifier(text[:1024], candidate_labels=ZS_LABELS, multi_label=False)
    out = {c: 0.0 for c in VALID_CONDITIONS}
    for lbl, score in zip(result['labels'], result['scores']):
        cond = ZS_TO_CONDITION.get(lbl)
        if cond and cond in out:
            out[cond] += float(score)
    total = sum(out.values()) or 1.0
    return {k: v / total for k, v in out.items()}


def apply_keyword_rules(blended, text):
    """Boost specific conditions based on keyword presence in text."""
    text_lower = text.lower()
    for keywords, condition, multiplier in KEYWORD_RULES:
        if any(kw in text_lower for kw in keywords):
            if condition in blended:
                blended[condition] *= multiplier

    total = sum(blended.values()) or 1.0
    return {k: v / total for k, v in blended.items()}


def schema_to_idx(label):
    classes = list(le.classes_)
    return classes.index(label) if label in classes else None


def classify_batch(texts, threshold=CONFIDENCE_THRESHOLD):
    your_proba = run_your_model(texts)
    rows, blended_rows = [], []
    print(f'   Classifying {len(texts)} participants...')

    for i, text in enumerate(texts):
        text_lower = text.lower()
        text_words = set(re.sub(r"[^a-z ]", "", text_lower).split())

        your_dict = {
            le.classes_[j]: float(your_proba[i][j]) for j in range(len(le.classes_))
        }

        your_valid = {c: your_dict.get(c, 0.0) for c in VALID_CONDITIONS}
        t = sum(your_valid.values()) or 1.0
        your_valid = {k: v / t for k, v in your_valid.items()}

        zs_dict = run_zeroshot(text)

        blended = {
            c: REDDIT_WEIGHT * your_valid.get(c, 0.0)
            + BART_WEIGHT * zs_dict.get(c, 0.0)
            for c in VALID_CONDITIONS
        }

        blended = apply_keyword_rules(blended, text)

        sorted_blended = sorted(blended.items(), key=lambda x: -x[1])
        final_label = sorted_blended[0][0]
        final_conf = sorted_blended[0][1]

        your_top = max(your_valid, key=your_valid.get)
        zs_top = max(zs_dict, key=zs_dict.get)
        agree = your_top == zs_top

        has_distress = bool(text_words & DISTRESS_INDICATORS)
        has_normal = bool(text_words & NORMAL_INDICATORS)

        if final_label == 'suicidewatch':
            risk_level, status = 'high', 'condition_detected'
        elif not has_distress and has_normal and final_conf < BORDERLINE_LOWER + 0.05:

            final_label = 'normal'
            risk_level, status = 'low', 'normal'
        elif final_conf >= threshold:
            risk_level = 'high'
            status = 'condition_detected'
        elif final_conf >= BORDERLINE_LOWER:
            risk_level, status = 'borderline', 'borderline'
        else:
            final_label = 'normal'
            risk_level, status = 'low', 'normal'

        top5 = sorted_blended[:5]
        rows.append(
            {
                'participant_id': df['participant_id'].iloc[i],
                'classification': final_label,
                'confidence': round(final_conf, 4),
                'status': status,
                'risk_level': risk_level,
                'models_agree': agree,
                'your_model_top': your_top,
                'zeroshot_top': zs_top,
                'top1_condition': top5[0][0],
                'top1_prob': round(top5[0][1], 4),
                'top2_condition': top5[1][0] if len(top5) > 1 else '',
                'top2_prob': round(top5[1][1], 4) if len(top5) > 1 else 0.0,
                'word_count': df['word_count'].iloc[i],
            }
        )
        blended_full = {l: blended.get(l, 0.0) for l in VALID_CONDITIONS}
        blended_full['participant_id'] = df['participant_id'].iloc[i]
        blended_rows.append(blended_full)

        if (i + 1) % 5 == 0 or (i + 1) == len(texts):
            print(f'   ... {i+1}/{len(texts)} done')

    global blended_matrix_df
    blended_matrix_df = pd.DataFrame(blended_rows).set_index('participant_id')
    return pd.DataFrame(rows)


print('\n✅ Ensemble ready  —  mentalhealth removed from output')
print('   Valid output conditions:', ', '.join(VALID_CONDITIONS))
```

```python
print('🔬 Chart 7 — SHAP Global XAI')
print('=' * 70)

X_shap = tfidf.transform(texts_list)
feature_names = np.array(tfidf.get_feature_names_out())


from sklearn.feature_extraction.text import ENGLISH_STOP_WORDS

EXTRA_NOISE = {
    've',
    'll',
    'didn',
    'don',
    'isn',
    'wasn',
    'aren',
    'wouldn',
    'couldn',
    'shouldn',
    'doesn',
    'hadn',
    'hasn',
    'won',
    're',
    'just',
    'like',
    'really',
    'actually',
    'going',
    'getting',
    'things',
    'thing',
    'know',
    'think',
    'way',
    'want',
    'need',
    'feel',
    'feeling',
    'felt',
    'make',
    'made',
    'time',
    'day',
    'people',
    'person',
    'got',
    'get',
    'did',
    'does',
    'im',
    'ive',
    'id',
    'doesnt',
    'didnt',
    'cant',
    'wont',
    'ive',
}
ALL_NOISE = ENGLISH_STOP_WORDS | EXTRA_NOISE


def is_real_word(token):

    if not token.isalpha():
        return False

    if len(token) < 3:
        return False

    if token.lower() in ALL_NOISE:
        return False
    return True


real_word_mask = np.array([is_real_word(f) for f in feature_names])
real_word_indices = np.where(real_word_mask)[0]
print(
    f'   ✅ {len(real_word_indices)} / {len(feature_names)} features are real words (filtered {len(feature_names)-len(real_word_indices)} noise tokens)'
)

print('⚙️  Computing SHAP values...')
shap_explainer = shap.LinearExplainer(lr, X_shap, feature_perturbation='interventional')
print('⚙️  Computing SHAP values...')
shap_explainer = shap.LinearExplainer(lr, X_shap, feature_perturbation='interventional')
shap_vals = shap_explainer.shap_values(X_shap)


if isinstance(shap_vals, np.ndarray) and shap_vals.ndim == 3:

    shap_vals = [shap_vals[:, :, c] for c in range(shap_vals.shape[2])]
    print(f'   ℹ️  New SHAP API detected — reshaped to {len(shap_vals)} class arrays')

print(f'   ✅ Done — {X_shap.shape[0]} participants × {len(shap_vals)} classes')


detected_schema = (
    results_df[results_df['status'] == 'condition_detected']['classification']
    .unique()
    .tolist()
)
if not detected_schema:
    detected_schema = results_df['classification'].value_counts().head(5).index.tolist()

condition_indices = [
    i for i in [schema_to_idx(s) for s in detected_schema] if i is not None
]
if not condition_indices:
    condition_indices = list(range(min(5, len(shap_vals))))


mean_abs_shap = np.mean(
    [np.abs(shap_vals[i]).mean(axis=0) for i in condition_indices], axis=0
)
filtered_shap = mean_abs_shap[real_word_indices]
top20_local = np.argsort(filtered_shap)[::-1][:20]
top20_global_idx = real_word_indices[top20_local]
top20_words = feature_names[top20_global_idx]
top20_vals = filtered_shap[top20_local]

print(f'\n   📋 Top 20 meaningful words: {", ".join(top20_words[:10])}...')

fig, axes = plt.subplots(1, 2, figsize=(20, 8))
fig.suptitle(
    'SHAP Global XAI — Most Influential Words (Survey Participants)',
    fontsize=14,
    fontweight='bold',
)

ax = axes[0]
colors_shap = plt.cm.RdYlGn_r(np.linspace(0.1, 0.9, len(top20_words)))
ax.barh(top20_words[::-1], top20_vals[::-1], color=colors_shap[::-1])
ax.set_xlabel(
    'Mean |SHAP Value|  (higher = more influential for prediction)', fontsize=10
)
ax.set_title(
    'Top 20 Meaningful Words\n(filtered: numbers/dosages/noise removed)',
    fontweight='bold',
    fontsize=11,
)
ax.tick_params(axis='y', labelsize=9)
ax.text(
    0.98,
    0.02,
    f'Based on {X_shap.shape[0]} survey responses',
    transform=ax.transAxes,
    ha='right',
    fontsize=8,
    color='gray',
)

ax = axes[1]
top_classes = detected_schema[:6]
plot_data = {}
for cname in top_classes:
    cidx = schema_to_idx(cname)
    if cidx is None:
        continue

    mean_abs_cond = np.abs(shap_vals[cidx]).mean(axis=0)
    filtered_cond = mean_abs_cond[real_word_indices]
    top3_local = np.argsort(filtered_cond)[::-1][:3]
    top3_global = real_word_indices[top3_local]
    for fi in top3_global:
        word = feature_names[fi]
        val = mean_abs_cond[fi]
        plot_data[f'{word}\n({cname})'] = (val, cname)

if plot_data:
    sorted_pd = sorted(plot_data.items(), key=lambda x: -x[1][0])
    words_pd = [s[0] for s in sorted_pd]
    vals_pd = [s[1][0] for s in sorted_pd]
    cnames_pd = [s[1][1] for s in sorted_pd]
    ax.barh(
        words_pd,
        vals_pd,
        color=[CLASS_COLOR.get(c, '#aaa') for c in cnames_pd],
        alpha=0.85,
        edgecolor='white',
    )
    ax.set_xlabel('Mean |SHAP Value|', fontsize=10)
    ax.set_title(
        'Top 3 Key Words per Detected Condition\n(colour = condition)',
        fontweight='bold',
        fontsize=11,
    )
    ax.legend(
        handles=[
            mpatches.Patch(color=CLASS_COLOR.get(c, '#aaa'), label=c)
            for c in top_classes
        ],
        fontsize=8,
        loc='lower right',
    )
    ax.tick_params(axis='y', labelsize=8)

plt.tight_layout()
plt.savefig(_paths.output_path('chart7_shap_global.png'), dpi=150, bbox_inches='tight')
plt.show()
print('✅ Saved → chart7_shap_global.png')
```

```python
print('🔬 Chart 8 — SHAP Per-Condition Deep Dive')
print('=' * 70)


detected_conditions = (
    results_df[results_df['status'] == 'condition_detected']['classification']
    .value_counts()
    .head(4)
    .index.tolist()
)
if not detected_conditions:
    detected_conditions = (
        results_df['top1_condition'].value_counts().head(3).index.tolist()
    )

n_plots = len(detected_conditions)
if n_plots == 0:
    print('⚠️  No detected conditions — run Cell 7 first.')
else:
    fig, axes = plt.subplots(1, n_plots, figsize=(6 * n_plots, 8), squeeze=False)
    axes = axes[0]
    fig.suptitle(
        'SHAP Per-Condition: Which Words Drive Each Classification',
        fontsize=13,
        fontweight='bold',
    )

    for ax, cname in zip(axes, detected_conditions):
        cidx = schema_to_idx(cname)
        if cidx is None:
            ax.text(0.5, 0.5, f'{cname}\nnot in encoder', ha='center', va='center')
            ax.axis('off')
            continue

        sv = shap_vals[cidx]

        sv_filtered = sv[:, real_word_indices]
        words_filtered = feature_names[real_word_indices]

        pos_mean = np.where(sv_filtered > 0, sv_filtered, 0).mean(axis=0)
        neg_mean = np.where(sv_filtered < 0, sv_filtered, 0).mean(axis=0)

        top_pos = list(np.argsort(pos_mean)[::-1][:8])
        top_neg = list(np.argsort(neg_mean)[:8])
        combined = list(dict.fromkeys(top_pos + top_neg))[:12]

        words_plot = words_filtered[combined]
        vals_plot = sv_filtered.mean(axis=0)[combined]

        sorted_pairs = sorted(zip(vals_plot, words_plot))
        vals_plot, words_plot = zip(*sorted_pairs)

        bar_colors = ['#e74c3c' if v > 0 else '#3498db' for v in vals_plot]
        ax.barh(words_plot, vals_plot, color=bar_colors, alpha=0.85, edgecolor='white')
        ax.axvline(0, color='black', lw=1)
        ax.set_title(
            f'[{cname.upper()}]',
            fontweight='bold',
            fontsize=11,
            color=CLASS_COLOR.get(cname, 'black'),
        )
        ax.set_xlabel('Mean SHAP Value', fontsize=9)
        n_in_cond = (results_df['classification'] == cname).sum()
        ax.set_xlabel(f'Mean SHAP Value  ({n_in_cond} participants)', fontsize=9)
        ax.legend(
            handles=[
                mpatches.Patch(color='#e74c3c', label='↑ Pushes TOWARD this condition'),
                mpatches.Patch(
                    color='#3498db', label='↓ Pushes AWAY from this condition'
                ),
            ],
            fontsize=7,
            loc='lower right',
        )
        ax.tick_params(axis='y', labelsize=9)

    plt.tight_layout()
    plt.savefig(
        _paths.output_path('chart8_shap_per_condition.png'),
        dpi=150,
        bbox_inches='tight',
    )
    plt.show()
    print('✅ Saved → chart8_shap_per_condition.png')

    print('\n📋 SEMANTIC SUMMARY ')
    print('=' * 70)
    for cname in detected_conditions:
        cidx = schema_to_idx(cname)
        if cidx is None:
            continue
        sv_filtered = shap_vals[cidx][:, real_word_indices]
        pos_mean = np.where(sv_filtered > 0, sv_filtered, 0).mean(axis=0)
        top_words = feature_names[real_word_indices][np.argsort(pos_mean)[::-1][:8]]
        n_in_cond = (results_df['classification'] == cname).sum()
        print(f'\n  [{cname.upper()}]  ({n_in_cond} participant(s))')
        print(f'     Key words : {", ".join(top_words)}')
        print(f'     Meaning   : These are the words in participant responses that')
        print(f'                 most strongly push the model toward [{cname}].')
        print(f'                 Higher = stronger signal for this condition.')
```

## 📄 Research Paper — Methods Snippet

> *Text responses were collected via a Google Form survey and analysed using a two-model ensemble. 
A Reddit-trained TF-IDF ensemble (Logistic Regression, Linear SVM, Random Forest) provided 
clinical-vocabulary signal with class-prior bias correction. 
A zero-shot NLI classifier (facebook/bart-large-mnli; Lewis et al., 2020) provided domain-agnostic 
semantic classification using 15 survey-calibrated candidate labels mapped directly to the Reddit 
mental health taxonomy. Final probabilities were determined by a weighted blend (20% TF-IDF, 80% NLI), 
with a keyword-based crisis escalation override for suicidal ideation and a distress-vocabulary-based 
confidence adjustment. Classification thresholds were calibrated to survey language (condition: 0.20, 
borderline: 0.12). SHAP LinearExplainer (Lundberg & Lee, 2017) was applied to the Logistic Regression 
for explainability. All analysis was performed in Python 3.10 using scikit-learn, HuggingFace Transformers, and SHAP.*

**References:**
- Lewis et al. (2020). BART: Denoising Sequence-to-Sequence Pre-training. *ACL 2020*.
- Lundberg & Lee (2017). A unified approach to interpreting model predictions. *NeurIPS 2017*.
- Pedregosa et al. (2011). Scikit-learn: Machine Learning in Python. *JMLR 12*.
